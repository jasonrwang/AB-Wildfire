An interactive companion to `AB-Wildfire.ipynb`, `BC-Wildfire.ipynb`, `CA-Wildfire.ipynb`, and `Summary Stats.ipynb`, built with [Altair](https://altair-viz.github.io/) so the main charts from this project can be explored directly rather than only viewed as static PNGs.

The same charts are published at https://jasonrwang.github.io/AB-Wildfire/. This notebook reads the compiled CSVs produced by the other notebooks (`data/ABWildfires.csv`, `data/BCWildfires.csv`, `data/CAWildfires.csv`) — run those first if you want to regenerate the compiled data from raw sources.

**How to interact:**
- **Chart 1** — drag across the small bar chart at the bottom to zoom the detail chart above into a year range.
- **Chart 2** — click a province in the legend to isolate it (click the background to reset).
- **Chart 3** — click a province in the legend to isolate it, and use the radio buttons to switch between raw area burned and area normalized by province size.


```python
import altair as alt

from dashboard import charts

alt.data_transformers.enable('default')

# Chart definitions live in dashboard/charts.py so this notebook and the
# GitHub Pages build (dashboard/build.py) render identical charts.
data = charts.load_data()
df_ab_yearly = charts.yearly_totals(data['AB'])
df_bc_yearly = charts.yearly_totals(data['BC'])
```

## 1. Alberta: Area Burned by Wildfire, 1918–present

Bar height is total hectares burned each year (the current season is year-to-date, not final); the orange line is the trailing 5-year average. The most recent year is highlighted in red.

```python
charts.ab_chart(df_ab_yearly)
```

## 2. Alberta vs. British Columbia

Stacked yearly area burned for both provinces, through the current season year-to-date (BC data begins in 1917/1918).

```python
charts.ab_bc_chart(df_ab_yearly, df_bc_yearly)
```

## 3. Canada-wide by Province, 1946–present

All provinces/territories from the Canadian National Fire Database (fires ≥ 200 ha only). Switch between raw area burned and area normalized by each province's land area (share of provincial land area burned) using the radio buttons below the chart.

```python
charts.ca_chart(data['CA'])
```

## 4. Summary: Alberta Area Burned by Decade

Total, mean, and maximum single-year area burned per decade, for quick comparison across eras.

```python
charts.decade_chart(df_ab_yearly)
```
